# Entropy Billiards: Visualising Uncertainty Through Physics

An interactive demonstration of entropy increase and thermodynamic equilibrium using billiards simulations.

## What is Entropy?

Entropy is a measure of disorder or uncertainty in a system. In this simulation, we'll observe how:

- **Ordered states** (particles aligned, moving together) spontaneously evolve into **disordered states** (particles scattered, random velocities)
- This is the **Second Law of Thermodynamics** in action: entropy tends to increase over time
- The velocity distribution evolves toward the **Maxwell-Boltzmann distribution** (thermodynamic equilibrium)

## The Demonstration

We'll simulate elastic collisions between particles in a 2D box. Watch how an initially ordered configuration (hot/cold regions or uniform alignment) quickly becomes disordered through collisions!

**Interactive controls allow you to:**
- Choose different starting configurations
- Control the number of particles
- Show/hide velocity arrows
- Adjust animation speed


## Setup

First, let's import the necessary modules and configure the notebook for interactive visualisation.


```python
# Enable interactive matplotlib in Jupyter
%matplotlib widget

import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
import ipywidgets as widgets
from IPython.display import display, HTML

# Import Kappenball modules
from kappenball import billiards, visualisation

print("✅ Modules loaded successfully!")
print("📊 Ready to explore entropy and uncertainty...")
```

## Quick Static Example

Let's start with a simple static visualisation to see what the simulation looks like.


```python
# Create initial state with hot/cold regions
state = billiards.initialise(
    n_particles=20,
    init_type='hotCold',  # Hot particles (yellow) on top, cold (cyan) on bottom
    box_xlim=(-2, 2),
    box_ylim=(-2, 2),
    random_state=42
)

# Plot the initial configuration
fig, ax = visualisation.plot_billiards_static(
    state,
    show_velocities=True,
    title="Initial State: Hot (Yellow) and Cold (Cyan) Particles"
)
plt.show()

print("\n🔥 Yellow particles = HOT (fast moving)")
print("❄️  Cyan particles = COLD (slow moving)")
print("⏩ Arrows show velocity direction and magnitude")
```

## Observe Entropy Increase

Let's run the simulation and compare the initial ordered state with the evolved disordered state.


```python
# Simulate for 500 steps
state_initial = billiards.initialise(
    n_particles=20,
    init_type='hotCold',
    box_xlim=(-2, 2),
    box_ylim=(-2, 2),
    random_state=42
)

state_evolved = state_initial.copy()
for _ in range(500):
    state_evolved = billiards.simulate_step(state_evolved, dt=0.002)

# Plot side by side
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))

# Initial state
ax1.set_xlim(state_initial['box_xlim'])
ax1.set_ylim(state_initial['box_ylim'])
ax1.set_aspect('equal')
ax1.set_facecolor('#4080CC')
ax1.set_title('t = 0: Ordered (Low Entropy)', fontsize=14, fontweight='bold')
visualisation.plot_billiards_frame(ax1, state_initial, show_velocities=True)

# Evolved state
ax2.set_xlim(state_evolved['box_xlim'])
ax2.set_ylim(state_evolved['box_ylim'])
ax2.set_aspect('equal')
ax2.set_facecolor('#4080CC')
ax2.set_title('t = 500 steps: Disordered (High Entropy)', fontsize=14, fontweight='bold')
visualisation.plot_billiards_frame(ax2, state_evolved, show_velocities=True)

plt.tight_layout()
plt.show()

print("\n🔄 Notice how:")
print("   • Particles have mixed (no longer separate hot/cold regions)")
print("   • Velocities are more randomly distributed")
print("   • The system has moved toward equilibrium")
print("\n🎲 This is entropy increase in action!")
```

## Interactive Animation

Now watch the entropy increase in real-time! Run the cell below to create an animation.


```python
print("🎬 Creating animation (this may take a moment)...")

# Create initial state
state = billiards.initialise(
    n_particles=20,
    init_type='hotCold',
    box_xlim=(-2, 2),
    box_ylim=(-2, 2),
    random_state=None  # Random each time
)

# Create and display animation
fig, anim = visualisation.animate_billiards(
    state,
    n_frames=300,
    dt=0.002,
    interval=20,  # 20ms = 50 fps
    show_velocities=True,
    title="Entropy Billiards: Watch Order → Disorder"
)

plt.show()
print("\n✅ Watch the hot (yellow) and cold (cyan) particles mix!")
print("🎯 Observe entropy increasing as order becomes disorder.")
```

## Velocity Distribution Analysis

Let's examine how the velocity distribution evolves toward the Maxwell-Boltzmann distribution.


```python
print("📊 Analysing velocity distribution evolution...\n")

# Run simulation and collect states
state = billiards.initialise(
    n_particles=30,
    init_type='hotCold',
    box_xlim=(-2, 2),
    box_ylim=(-2, 2),
    random_state=42
)

states = [state.copy()]
for _ in range(2000):
    state = billiards.simulate_step(state, dt=0.001)
    if _ % 20 == 0:  # Save every 20th state
        states.append(state.copy())

# Plot velocity histogram
fig, ax = visualisation.plot_velocity_histogram(
    states,
    bins=40,
    title="Velocity Distribution: Evolution to Equilibrium"
)
plt.show()

print("\n🎯 Key observations:")
print("   • Distribution becomes smoother over time")
print("   • System approaches Maxwell-Boltzmann distribution")
print("   • This is thermalization: reaching equilibrium!")
```

## Energy Conservation

Let's verify that our simulation conserves energy (as it should for elastic collisions).


```python
print("⚡ Checking energy conservation...\n")

# Run simulation
state = billiards.initialise(
    n_particles=15,
    init_type='rand',
    box_xlim=(-2, 2),
    box_ylim=(-2, 2),
    random_state=42
)

states = [state.copy()]
for _ in range(1000):
    state = billiards.simulate_step(state, dt=0.001)
    states.append(state.copy())

# Plot energy evolution
fig, ax = visualisation.plot_energy_evolution(
    states,
    title="Total Kinetic Energy vs Time"
)
plt.show()

# Calculate energy statistics
energies = []
for s in states:
    E = 0.5 * np.sum(s['mass'][:, np.newaxis] * s['V']**2)
    energies.append(E)

mean_E = np.mean(energies)
std_E = np.std(energies)

print(f"\n📈 Energy Statistics:")
print(f"   Mean Energy: {mean_E:.6e}")
print(f"   Std Dev: {std_E:.6e}")
print(f"   Variation: {100*std_E/mean_E:.2f}%")
print("\n✅ Energy is approximately conserved!")
print("   (Small variations due to numerical integration)")
```
