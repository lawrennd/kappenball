# Kappenball: Interactive Uncertainty Game

**Demonstrating Uncertainty Through Play**

Welcome to Kappenball! This interactive game demonstrates fundamental concepts of uncertainty, probability, and control theory through a simple but engaging physics simulation.

## The Game

A ball falls from the top of the screen through a field of pins. Your goal is to guide the ball through one of two holes in the pin array using keyboard controls (left/right arrows). But there's a catch: the ball's motion is subject to random uncertainty!

## Learning Objectives

1. **Uncertainty**: Experience how randomness affects deterministic systems
2. **Control**: Learn about the trade-off between control effort and success
3. **Probability**: Observe probabilistic outcomes in real-time
4. **Energy**: Understand the cost of controlling uncertain systems

## How It Works

- The ball starts at the top centre and falls downward
- Random horizontal velocity noise is added at each step (adjustable)
- You can apply left/right control inputs (costs energy)
- Success: ball passes through a hole (score increases)
- Failure: ball hits pins ("BANG!" and reset)

## Key Metrics

- **Score**: Number of successful drops
- **Energy**: Total control inputs used
- **Average**: Energy per successful drop (efficiency)
